```python
def card_points(value):
  """Return the points of one card value."""
  return points.get(value, value)
def hand_total(hand):
  """Return the total points of a list of card values."""
  total = 0
  for value in hand:
    total += card_points(value)
  return total
```

```python
card_points("King") #10
```

```python
card_points(7)
```

```python
hand_total(["Ace", 7, "Queen"]) #28
```

```python
hand_total() #Error
```

```python
def deal(n):
    hand = []
    for i in range(n):
        card = (random.choice(values), random.choice(suits))
        hand.append(card)
    return hand

hand1 = deal(5)
hand2 = deal(5)

values1 = [value for value, suit in hand1]
values2 = [value for value, suit in hand2]

print("Hand 1:", hand1)
print("Total:", hand_total(values1))

print("Hand 2:", hand2)
print("Total:", hand_total(values2))
```

PART D

```python
gradebook = {}
def add_student(book, name, mark):
    """Adds or updates a student's mark."""
    book[name] = mark


def average(book):
    """Returns the average mark."""
    return sum(book.values()) / len(book)


def top_student(book):
    """Returns the name of the student with the highest mark."""
    return max(book, key=book.get)


def passed(book):
    """Returns a list of students with marks of 50 or more."""
    return [name for name, mark in book.items() if mark >= 50]


add_student(gradebook, "Ana", 88)
add_student(gradebook, "Ben", 45)
add_student(gradebook, "Cleo", 95)
add_student(gradebook, "Dev", 72)
add_student(gradebook, "Eli", 30)

print(f"Average: {average(gradebook):.1f}")
print("Top student:", top_student(gradebook))
print("Passed:", passed(gradebook))
```
