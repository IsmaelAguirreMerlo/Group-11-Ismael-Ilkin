```python
import random
suits=["Hearts", "Diamonds", "Clubs", "Spades" ]
values= [ "Ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "Jack", "Queen", "King"]
print(random.choice(values), "of", random.choice(suits))
print(random.choice(values), "of", random.choice(suits))
print(random.choice(values), "of", random.choice(suits))
print("Cards in a full deck:", len(suits) * len(values))
```
