Check Point 2

```python
suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
values = ["Ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "Jack", "Queen", "King"]


deck = []
for suit in suits:
  for value in values:
    deck.append((value, suit))
import random
random.shuffle(deck)
hand = deck[0:5]
for value, suit in hand:
    print(f"{value} of {suit}")
```
