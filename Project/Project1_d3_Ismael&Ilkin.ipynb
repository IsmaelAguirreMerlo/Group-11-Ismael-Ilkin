**Project 1-Kitchen Inventory**

```python
inventory = {}
```

```python
def add_ingredient(name, quantity, unit):
      if name not in inventory:
        inventory[name]= {"quantity":quantity, "unit": unit}
      else:
        print("Ingredient already exists")

q1=input("Do you want to add product(s)?")
if q1=="yes":
  np=int(input("Number of products:"))
  for i in range (np):
    name=input(f"\nProduct {i+1}: ")
    quantity=int(input("Quantity: "))
    unit=input("Unit: ")
    add_ingredient(name, quantity, unit)

def view_inventory():
  for name, info in inventory.items():
    print("-",name, info["quantity"],info["unit"])

q2=input("\nView inventory(yes/no): ")
if q2=="yes":
  view_inventory()


```

```python
recipes={}
```

```python
def add_recipe(rname, ingredients):
  if rname not in recipes:
    recipes[rname]=ingredients
  else:
    print("Recipe already exists")

def add_ingredients(ingredient, quant, unt):
      if ingredient not in ingredients:
        ingredients[ingredient]= {"quantity":quant, "unit": unt}
      else:
        print("Ingredient already exists")

while True:
  q3=input("\nDo you want to add a recipe (yes/no): ")
  if q3=="no":
    break

  if q3 == "yes":
    rname=input("\nRecipe name: ")
    ingredients={}
    c=1
    while True:
      ing=input(f"Ingrediente {c} or 'done': ")
      if ing =="done":
        break

      quant=int(input("Quantity: "))
      unt=input("Unit: ")
      add_ingredients(ing, quant, unt)
      c+=1

    add_recipe(rname, ingredients)



def view_recipes():
  for rname, ingredients in recipes.items():
    print("\n---",rname, "---")
    for ingredient, info in ingredients.items():
      print("-",ingredient, info["quantity"], info["unit"])


q4=input("\nView recipes(yes/no): ")
if q4=="yes":
  view_recipes()


```

```python
recipe=input("Which recipe do you want to check? ")
if recipe in recipes:
  print("\n---",recipe, "---")
  for ingredient, info in recipes[recipe].items():
    print("\n-",ingredient, info["quantity"], info["unit"])
    if ingredient not in inventory:
      print(f"{ingredient} is missing")
    elif inventory[ingredient]["quantity"] >= info["quantity"]:
      print(f"There is enough {ingredient}")
    else:
      print(f"You need {info['quantity']-inventory[ingredient]['quantity']} {info['unit']} of {ingredient}")



```
