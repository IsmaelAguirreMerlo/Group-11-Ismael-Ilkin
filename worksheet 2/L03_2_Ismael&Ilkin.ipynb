```python
import random

suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
values = ["Ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "Jack", "Queen", "King"]

deck = []

for suit in suits:
    for value in values:
        deck.append((value, suit))

random.shuffle(deck)

hand = deck[:5]

print("Your hand:")

for value, suit in hand:
    print(f"{value} of {suit}")

hand_suits = []

for value, suit in hand:
    hand_suits.append(suit)

hearts = hand_suits.count("Hearts")
print(f"\nNumber of Hearts: {hearts}")

picture_cards = 0

for value, suit in hand:
    if value in ["Jack", "Queen", "King"]:
        picture_cards += 1

print(f"Number of picture cards: {picture_cards}")

if hand_suits.count(hand_suits[0]) == 5:
    print("Flush!")
else:
    print("Not a flush")
```

```python
import random, time
from IPython.display import clear_output, display, HTML

WIDTH, HEIGHT = 70, 22        # columns and rows of characters
TRAIL = 10                    # length of each falling trail
symbols = "01{}()=;<>[]+*#$%&"

# one falling "drop" per column; its value is the row of the bright head
drops = [random.randint(-HEIGHT, 0) for col in range(WIDTH)]

for frame in range(300):      # 300 frames = about 15 seconds
    rows = []
    for row in range(HEIGHT):
        line = ""
        for col in range(WIDTH):
            distance = drops[col] - row
            if distance == 0:
                line += '<span style="color:#d6ffd6">' + random.choice(symbols) + '</span>'
            elif 0 < distance < TRAIL:
                fade = 1 - distance / TRAIL
                line += f'<span style="color:rgba(0,255,65,{fade:.2f})">' + random.choice(symbols) + '</span>'
            else:
                line += " "
        rows.append(line)

    clear_output(wait=True)
    display(HTML('<pre style="background:#000;font-size:14px;line-height:1.1;padding:10px">'
                 + "\n".join(rows) + "</pre>"))

    for col in range(WIDTH):
        drops[col] += 1
        if drops[col] - TRAIL > HEIGHT and random.random() < 0.1:
            drops[col] = 0     # start a new drop at the top
    time.sleep(0.05)
```
